```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
k = 5
A = np.eye(k)
A[-1, :] = -1
A[-1, -1] = 0
A
```

```python
U, S, Vt = np.linalg.svd(A)
U, S, Vt
```

```python
Q, R = np.linalg.qr(A)
Q, R
```

```python
np.round(Q @ Q.T), np.round(Q.T @ Q)
```

```python
Qr = Q[:, :-1]
Qr
```

```python
np.round(Qr @ Qr.T, 4), np.round(Qr.T @ Qr, 4)
```

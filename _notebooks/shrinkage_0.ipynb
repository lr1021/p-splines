```python
x = np.linspace(-3, 3, 100)
# f = np.minimum(np.ones_like(x), np.exp((1.0 - x)))
f = np.tanh((1.0 - x)*3)/2+1/2
plt.plot(x, f)
```

```python
scale = 4.0
alt_data = {('f6b', 0.1): {}}
for r in data[('f6b', 0.1)].keys():
    x, y = data[('f6b', 0.1)][r]
    x = x.copy()
    y = y.copy()
    f = np.tanh((1.0 - x)*3)/2+1/2

    y *= (1+scale*f)
    alt_data[('f6b', 0.1)][r] = (x, y)
```
